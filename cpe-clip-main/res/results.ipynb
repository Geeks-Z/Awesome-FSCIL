```python
import json

def get_eval_accuracies(file):
    """提取所有评估阶段的准确率（Eval/Exp*），并转换为百分比格式"""
    with open(file) as f:
        data = json.load(f)
    d = {}
    for k, v in data['Acc'].items():
        if len(k.split('/'))==3:
            d[k] = v
    return {k:v*100 for k,v in zip(range(len(d)), d.values())}

def calculate_metrics(files):
    """计算三组文件的指标：首项平均、末项平均、组平均值的平均"""
    all_data = [get_eval_accuracies(f) for f in files]
    
    # 提取每组第零个索引和最后一个索引
    first_items = [data[0] for data in all_data]  
    last_items = [data[max(data.keys())] for data in all_data]  # 最后一个任务
    
    # 计算每组平均值
    group_means = [sum(data.values()) / len(data) for data in all_data]
    
    # 保留两位小数
    results = {
        "first_items_avg": round(sum(first_items) / len(first_items), 2),
        "last_items_avg": round(sum(last_items) / len(last_items), 2),
        "group_means_avg": round(sum(group_means) / len(group_means), 2),
        "raw_first_items": [round(x, 2) for x in first_items],
        "raw_last_items": [round(x, 2) for x in last_items],
        "raw_group_means": [round(x, 2) for x in group_means]
    }
    return results
def print_results(results):
    """格式化输出结果，添加换行符"""
    print(f"首项平均值 (first_items_avg): {results['first_items_avg']}\n")
    print(f"末项平均值 (last_items_avg): {results['last_items_avg']}\n")
    print(f"组平均值的平均 (group_means_avg): {results['group_means_avg']}\n")
    print(f"原始首项数据 (raw_first_items): {results['raw_first_items']}\n")
    print(f"原始末项数据 (raw_last_items): {results['raw_last_items']}\n")
    print(f"原始组平均值 (raw_group_means): {results['raw_group_means']}")

# 示例调用
files = [
    "temp_model_cifar100_Lg2_Dg12_TRMreplace_VRMaccumulate_run1.json",
    "temp_model_cifar100_Lg2_Dg12_TRMreplace_VRMaccumulate_run2.json",
    "temp_model_cifar100_Lg2_Dg12_TRMreplace_VRMaccumulate_run3.json"
]
results = calculate_metrics(files)
print_results(results)
```

```python
import json
import numpy as np

def get_eval_accuracies(file):
    """提取所有评估阶段的准确率（Eval/Exp*），并转换为百分比格式"""
    with open(file) as f:
        data = json.load(f)
    d = {}
    for k, v in data['Acc'].items():
        if len(k.split('/'))==3:
            d[k] = v
    return {k:v*100 for k,v in zip(range(len(d)), d.values())}

def calculate_metrics(files):
    """计算三组文件的指标：首项平均、末项平均、组平均值的平均，以及它们的方差"""
    all_data = [get_eval_accuracies(f) for f in files]
    
    # 提取每组第零个索引和最后一个索引
    first_items = [data[0] for data in all_data]  
    last_items = [data[max(data.keys())] for data in all_data]  # 最后一个任务
    
    # 计算每组平均值
    group_means = [sum(data.values()) / len(data) for data in all_data]
    
    # 计算均值和方差
    results = {
        # 均值部分
        "first_items_avg": round(np.mean(first_items), 2),
        "last_items_avg": round(np.mean(last_items), 2),
        "group_means_avg": round(np.mean(group_means), 2),
        
        # 方差部分
        "first_items_var": round(np.var(first_items), 2),
        "last_items_var": round(np.var(last_items), 2),
        "group_means_var": round(np.var(group_means), 2),
        
        # 原始数据
        "raw_first_items": [round(x, 2) for x in first_items],
        "raw_last_items": [round(x, 2) for x in last_items],
        "raw_group_means": [round(x, 2) for x in group_means]
    }
    return results

def print_results(results):
    """格式化输出结果，添加换行符"""
    print(f"Base {results['first_items_avg']}±{results['first_items_var']}")
    print(f"Last {results['last_items_avg']}±{results['last_items_var']}")
    print(f"Avg {results['group_means_avg']}±{results['group_means_var']}")
    
    print(f"原始首项数据 (raw_first_items): {results['raw_first_items']}\n")
    print(f"原始末项数据 (raw_last_items): {results['raw_last_items']}\n")
    print(f"原始组平均值 (raw_group_means): {results['raw_group_means']}")

# 示例调用
files = [
    "temp_model_cifar100_Lg2_Dg12_TRMreplace_VRMaccumulate_run1.json",
    "temp_model_cifar100_Lg2_Dg12_TRMreplace_VRMaccumulate_run2.json",
    "temp_model_cifar100_Lg2_Dg12_TRMreplace_VRMaccumulate_run3.json"
]
results = calculate_metrics(files)
print_results(results)
```
